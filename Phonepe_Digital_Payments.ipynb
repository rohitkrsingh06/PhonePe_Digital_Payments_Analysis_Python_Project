**TASK 3: DATA QUALITY CHECKS**

**--> Ensure data consistency across state and district levels.**

**--> Display the discripancies found between state-level and district-level data.**

```python
# STATE WISE AGGREGATE DATA FROM DISTRICT DATASET
result1=d_txnusers.groupby('State').agg({
    'Transactions':'sum',
    'Amount (INR)':'sum',
    'Registered Users':'sum'}).round(0).reset_index()

# STATE WISE AGGREGATE DATA FROM STATE DATASET
result2=s_txnusers.groupby('State').agg({
    'Transactions':'sum',
    'Amount (INR)':'sum',
    'Registered Users':'sum'}).round(0).reset_index()

# MERGE BOTH TABLES 
result=result1.merge(result2,on='State',suffixes=('_district','_state'))

# NOW WE WILL COMPUTE DISCREPANCIES
result['Transactions_diff']=result['Transactions_district']-result['Transactions_state']
result['Amount (INR)_diff']=result['Amount (INR)_district']-result['Amount (INR)_state']
result['Registered Users_diff']=result['Registered Users_district']-result['Registered Users_state']

# DISPLAY ONLY WITH DISCREPANCIES
discrepancies=result[(result['Transactions_diff']!=0) | (result['Amount (INR)_diff']!=0) | (result['Registered Users_diff']!=0)][
    ['State','Transactions_diff','Amount (INR)_diff','Registered Users_diff']]
print(discrepancies)
```

**INSIGHT--> There is a discrepancy in the data of state ANDHRA PRADESH in Amount(INR).**

**--> It looks like there are some missing data in in state data.**

**TASK 4: DATA MERGING AND ADVANCED ANALYSIS.**

**4.1: Ratio of Users to Population by state.**

```python
# AGGREGATE THE TOTAL POPULATION BY STATE 
res1=d_demo.groupby('State')['Population'].sum().reset_index()

# AGGREGATE THE TOTAL USERS BY STATE
res2=s_txnusers.groupby('State')['Registered Users'].max().reset_index()

# MERGE BOTH DATA

merged=res1.merge(res2,
                  on='State')
merged['Ratio_UtoP']=(merged['Registered Users']/merged['Population']).round(2)
print(merged)
```

```python
# PLOT A COLUMN CHART
plt.figure(figsize=(14,6))
sns.barplot(data=merged,x='State',y='Ratio_UtoP')
plt.title('RATIO OF USERS TO POPULATION BY STATE')
plt.xlabel('State')
plt.ylabel('User to Population Ratio')
plt.xticks(rotation=90)
plt.show()
```

**INSIGHT--> THE STATE OF DELHI HAS HIGHEST AND THE STATE OF MEGHALAYA HAS LOWEST USER TO POPULATION RATIO.**

**4.2--> Correlate population density with transaction volume.Calculate Correlation**

```python
# AGGREGATE THE TOTAL POPULATION DENSITY BY STATE
pdn=d_demo.groupby('State')['Density'].mean().reset_index()

# AGGREGATE THE TRANSACTIONS BY STATE
tst=s_txnusers.groupby('State')['Transactions'].sum().reset_index()

# MERGE BOTH DATASETS
merged=pdn.merge(tst,on='State')

# FIND CORRELATION BETWEEN DENSITY AND TRANSACTIONS
corr=merged['Density'].corr(merged['Transactions'])
print(f"The Correlation between Population Density and Transactions is: {corr}.")
```

**-->Create a scatter plot to visualize correlation.**

```python
# PLOT SCATTER PLOT
plt.figure(figsize=(10,6))
sns.scatterplot(data=merged,x='Density',y='Transactions')
plt.title('Population Density vs Transactions')
plt.xlabel('Population Density')
plt.ylabel('Transactions')
plt.ticklabel_format(style='plain',axis='y')
plt.tight_layout()
plt.show()
```

**INSIGHT:The scatter plot between Population Density and Transactions indicates a weak relationship between the two variables. While certain high-density regions demonstrate substantial transaction volumes, several lower-density regions also exhibit exceptionally high transaction activity. This suggests that transaction behavior is influenced by multiple economic and infrastructural factors beyond population concentration alone. Additionally, the presence of extreme outliers highlights significant regional variability in transaction patterns.**

**4.4--> Device Brand Usage Ratio.**

**-->Calculate the ratio of users using each device brand to the total number of registered users in each state.**

```python
# TAKE LATEST DATA FOR TOTAL REGISTERED USERS AND DEVICE TYPE USERS
latest_data_device=s_device[(s_device['Year']==latest_year) & (s_device['Quarter']==latest_quarter)]
latest_data_txn=s_txnusers[(s_txnusers['Year']==latest_year) & (s_txnusers['Quarter']==latest_quarter)]

# AGGREGATE
brand_users=latest_data_device.groupby(['State','Brand'])['Registered Users'].sum().reset_index()
total_users=latest_data_txn.groupby('State')['Registered Users'].sum().reset_index()
total_users=total_users.rename(columns={'Registered Users': 'Total_Users'})

# MERGE BOTH THE DATASET
merged=brand_users.merge(total_users,on='State')

# CALCULATE RATIO
merged['ratio']=(merged['Registered Users']/merged['Total_Users'])
merged['user_ratio%']=(merged['ratio']*100).round(2)
print(merged)
```

**--> Create a bar chart depicting the device brand usage ratio for each state.**

```python
# PLOT A BAR CHART FOR THE STATE OF MADHYA PRADESH
state_name = 'Madhya Pradesh'
plot_data = merged[merged['State'] == state_name]
plt.figure(figsize=(8,5))
ax=sns.barplot(
    data=plot_data,
    x='Brand',
    y='user_ratio%'
)
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%')     # To add data labels on top of bars in seaborn
plt.title(f'Device Brand Usage Ratio - {state_name}')
plt.xlabel('Brand')
plt.ylabel('Usage (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**INSIGHT--> DEVICE BRAND USAGE RATIO IS HIGHEST FOR XIAOMI BRAND.**
